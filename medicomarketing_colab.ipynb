```python
# @title 🔑 Step 2 — Enter your Anthropic API key
# Get one at platform.anthropic.com → API Keys (pay-as-you-go, ~$1-3 per full strategy run)

ANTHROPIC_API_KEY = "" # @param {type:"string"}

import os
os.environ["ANTHROPIC_API_KEY"] = ANTHROPIC_API_KEY

if ANTHROPIC_API_KEY.startswith("sk-ant-"):
    print("✅ API key saved for this session.")
else:
    print("⚠  This doesn't look right — the key should start with  sk-ant-")
```

```python
# @title 📋 Step 3a — Required fields

brand = "" # @param {type:"string"}
therapy_area = "" # @param {type:"string"}

if brand.strip() and therapy_area.strip():
    print(f"✅ Brand: {brand}  |  Therapy area: {therapy_area}")
else:
    print("⚠  Both fields are required — please fill them in.")
```

```python
# @title 📋 Step 3b — Optional details (leave blank to skip — more detail = better output)
# For list fields, separate items with a comma.

product       = "" # @param {type:"string"}
indication    = "" # @param {type:"string"}
market        = "" # @param {type:"string"}
business_goal = "" # @param {type:"string"}

# Comma-separated lists
target_specialties = "" # @param {type:"string"}
brand_evidence     = "" # @param {type:"string"}
hcp_insights       = "" # @param {type:"string"}
competitors        = "" # @param {type:"string"}
constraints        = "" # @param {type:"string"}

print("✅ Optional fields saved.")
```

```python
# @title ▶ Step 4 — Run the strategy
mode = "Quick test — Phase 1 only (~2-3 min)" # @param ["Quick test — Phase 1 only (~2-3 min)", "Full strategy — all 11 phases (~30-40 min)"]

# Build brief dict
brief = {"brand": brand.strip(), "therapy_area": therapy_area.strip()}

for key, val in [("product", product), ("indication", indication),
                 ("market", market), ("business_goal", business_goal)]:
    if val.strip():
        brief[key] = val.strip()

for key, val in [("target_specialties", target_specialties),
                 ("brand_evidence", brand_evidence),
                 ("hcp_insights", hcp_insights),
                 ("competitors", competitors),
                 ("constraints", constraints)]:
    items = [x.strip() for x in val.split(",") if x.strip()]
    if items:
        brief[key] = items

if not brief.get("brand") or not brief.get("therapy_area"):
    print("⚠  Please fill in Brand and Therapy area in Step 3a first.")
else:
    from medicomarketing_agent.engine import StrategyEngine
    from medicomarketing_agent.phases import PHASES

    phases_to_run = PHASES[:1] if "Quick" in mode else None
    engine = StrategyEngine(brief, out_dir="output")
    combined = engine.run_pipeline(phases_to_run)
    print(f"\n\n✅ Done! Run Step 5 to download your strategy.")
```

```python
# @title 📥 Step 5 — Download your results
import glob, os
from google.colab import files

output_files = sorted(glob.glob('output/*.md'))
if not output_files:
    print("No output files yet — run Step 4 first.")
else:
    print(f"Found {len(output_files)} file(s):")
    for f in output_files:
        print(f"  {f}")
    print("\nDownloading combined strategy document...")
    combined = 'output/medicomarketing-strategy.md'
    if os.path.exists(combined):
        files.download(combined)
    else:
        for f in output_files:
            files.download(f)
    print("Done. Check your browser downloads folder.")
```
